# Quaternion basics

Quick checks of the `Quaternion` class: axis-angle construction, products and rotation of a vector.

```python

from drone_sim.dynamics.quaternions import Quaternion
```

```python
import numpy as np
```

```python
q1 = Quaternion.from_axis_angle(np.array([1, 0, 0]), 90)
v = np.array([1,0,0])
rotated_v_in_q = q1.rotate(v)
```

```python
rotated_v_in_q.display()
```

```python
q2 = Quaternion(2,2,2,2)
q2 = q2.normalize()
q2.display()
```
